```python
from pathlib import Path
import pandas as pd

water_file = Path.cwd() / "Data" / "Water_Sample_Data.xlsx"
water_df = pd.read_excel(water_file, sheet_name="Data", header=[0, 1])

# Normalize the multi-level header for easier use.
water_df.columns = [
    col[0] if str(col[1]).startswith("Unnamed") else f"{col[0]} | {col[1]}"
    for col in water_df.columns
]

water_df.columns = [c.strip() for c in water_df.columns]

print(f"Loaded {water_df.shape[0]} rows and {water_df.shape[1]} columns")
water_df.head()
```
